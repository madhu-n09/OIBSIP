## Data Type Correction
The data types were verified and corrected to ensure consistency and improve the accuracy of future analysis.

```python
df.select_dtypes(include=["int64","float64"]).columns
```

```python
Q1=df["Age"].quantile(0.25)
Q3=df["Age"].quantile(0.75)
print("Q1:",Q1)
print("Q3:",Q3)
```

```python
IQR=Q3-Q1
print("IQR:",IQR)
```

```python
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR
print("Lower Limit:",lower)
print("Upper Limit:",upper)
```

```python
outliers=df[(df["Age"]<lower) | (df["Age"]>upper)]
print(outliers)
```

```python
Q1=df["LengthOfStay"].quantile(0.25)
Q3=df["LengthOfStay"].quantile(0.75)
IQR=Q3-Q1
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR
outliers=df[(df["LengthOfStay"]<lower) | (df["LengthOfStay"]>upper)]
print("Q1:",Q1,"\nQ3:",Q3,"\nIQR:",IQR,"\nLower Limit:",lower,"\nUpper Limit:",upper)
print(outliers)
```
